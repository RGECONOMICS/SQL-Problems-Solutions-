### 28) High freight charges - last year

<div style="background-color:#f5f5f5; border-left:5px solid #0071bc; padding:10px;">
<b>Task:</b> We're continuing to work on high freight charges. We now want to get the three ship countries with the highest average freight charges. But instead of filtering for a particular year, we want to use the last 12 months or order data, using as the end date the last OrderDate in Orders.
</div>



```python
query= """
SELECT 
    TOP 3 
    ShipCountry,
    ROUND(AVG(Freight),2) AS avg_freight
FROM Orders
WHERE OrderDate >= DATEADD(MONTH, -12, (SELECT MAX(OrderDate) FROM Orders))
GROUP BY ShipCountry
ORDER BY avg_freight DESC;
"""
df = pd.read_sql(query,engine)
df
```

### 29) Employee/Order Detail report

<div style="background-color:#f5f5f5; border-left:5px solid #0071bc; padding:10px;">
<b>Task:</b>We're doing inventory, and need to show Employee and Order Detail information like the below, for all orders. Sort by OrderID and Product.
</div>



```python
query= """
SELECT 
    e.EmployeeID, e.LastName, 
    o.OrderID, 
    p.ProductName, 
    od.Quantity
FROM Employees AS e

JOIN Orders AS o
ON e.EmployeeID = o.EmployeeID

JOIN OrderDetails AS od
ON o.OrderID = od.OrderID

JOIN Products AS p
ON p.ProductID = od.ProductID

ORDER BY o.OrderID, od.Quantity DESC 
"""
df = pd.read_sql(query,engine)
df
```

<div style="background-color:#e8f4fd; border-left:5px solid #0071bc; padding:10px;">
<b>Explanation:</b> This query connects multiple tables to display which employees handled each order and the products included in those orders. It joins the Employees, Orders, OrderDetails, and Products tables to show details such as employee name, order ID, product name, and quantity. Finally, the results are sorted by order ID and then by quantity in descending order to highlight the largest items per order.
</div>

### 30) Customers with no orders

<div style="background-color:#f5f5f5; border-left:5px solid #0071bc; padding:10px;">
<b>Task:</b> There are some customers who have never actually placed an order. Show these customers.

### 50. Customers with multiple orders in 5 day period.

<div style="background-color:#f5f5f5; border-left:5px solid #0071bc; padding:10px;">
<b>Task:</b> There are some customers for whom freight is a major expense when ordering from Northwind.However, by batching up their orders, and making one larger order instead of multiple smaller orders in a short period of time, they could reduce their freight costs significantly. Show those customers who have made more than 1 order in a 5 day period. The salespeople will use this to help customers reduce their freight costs.
</div>



```python
query= """
WITH NextOrders AS
(
    SELECT 
        CustomerID, CAST(OrderDate AS DATE) AS OrderDate,
        CAST(LEAD(OrderDate) OVER (PARTITION BY CustomerID ORDER BY CAST(OrderDate AS DATE) ASC) AS DATE) AS Next_Order_Date
    FROM Orders
)

SELECT
    CustomerID, OrderDate, Next_Order_Date,
     DATEDIFF(day, OrderDate, Next_Order_Date) AS Days_Between_Orders
FROM NextOrders
WHERE DATEDIFF(day, OrderDate, Next_Order_Date) <= 5
ORDER BY OrderDate, CustomerID, Days_Between_Orders;
"""
df = pd.read_sql(query,engine)
df
```

<b>Explanation:</b> This query finds customers who placed another order within 5 days of a previous one. It uses a CTE with the LEAD() function to get each customer's next order date, then calculates the days between orders. Only pairs with 5 or fewer days apart are shown.
